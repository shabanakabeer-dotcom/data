## UNIVARIATE ANALYSIS

```python
plt.figure(figsize=(6,4))
sns.histplot(pro["Age"],bins=10)
plt.show()
```

## Interpretation
#### - Customer ages are distributed fairly evenly between 18 and 70 years.
#### - No single age group dominates the data set.
#### - The data set contains customers from a wide range of age groups. 

```python
plt.figure(figsize=(8,5))
sns.countplot(x="Gender",data=pro)
plt.title("Distribution of Gender")
plt.xlabel("Gender")
plt.ylabel("count")
plt.xticks(rotation=45)
plt.show()
```

## Interprtation
#### - Male customers are significantly higher than female customers.
#### - Most purchases in the dataset are made by female customers.
#### - The dataset is not evenly distributed  across genders.
#### - Marketing strategies may focus more on male customers, as they represent the majority of shoppers


```python
plt.figure(figsize=(8,5))
sns.histplot(pro["Purchase Amount (USD)"])
plt.title("Distribution of Purchase Amount(USD)")
plt.xlabel("Purchase Amount (USD)")
plt.ylabel("Frequency")
plt.xticks(rotation=45)
plt.show()
```

## Interpretation
#### - Customers spend across all price ranges with no single amount dominating.
#### - The distribution is fairly uniform,indicating balanced spending behaviour.
#### - There are no extreme peakes or unusual putliers in the purchase amounts.

```python
plt.figure(figsize=(8,5))
sns.countplot(x="Category",data=pro)
plt.title("Distribution of Category")
plt.xlabel("Category")
plt.ylabel("count")
plt.xticks(rotation=45)
plt.show()
```

## Interpretation
#### - Clothing is the most sold/common category clearly dominate.
#### - Accessories is the second highest. 
#### - footwear and Outwear are much smaller categories.

## BIVARIATE ANALYSIS

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="Gender",y="Purchase Amount (USD)",data=pro)
plt.title("Gender vs Purchase Amount")
plt.xticks(rotation=45)
plt.show()
```

## Interpretation
#### - Male and Female spend almost the same amount o average.
#### - Males's average spend little more than Female's.
#### - Both groups are spending similar range of spending no big defferences.


```python
plt.figure(figsize=(8,5))
sns.scatterplot(x="Age",y="Purchase Amount (USD)",data=pro)
plt.title("Age vs Purchase Amount")
plt.xlabel("Age")
plt.ylabel("Purchase Amount (USD)")
plt.xticks(rotation=45)
plt.show()
```

## Interpretation
#### - points are spread all over the graph. No clear pattern age and purchase amount.
#### - Age does not stronglyaffect how much a person spends.
#### - No clear increases or decreasing trend. 

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="Category",y="Purchase Amount (USD)",data=pro)
plt.title("Category vs Purchase Amount")
plt.xticks(rotation=45)
plt.show()
```

## Interpretation
#### - Clothing buyers spend thd most on average.
#### - Outwear buyers spend the least on average.
#### - Footwear and Accessories show almost similar spending.

```python
plt.figure(figsize=(8,5))
sns.barplot(x="Subscription Status",y="Purchase Amount (USD)",data=pro)
plt.title("Age vs Purchase Amount")
plt.xlabel("Subscription Status")
plt.ylabel("Purchase Amount (USD)")
plt.xticks(rotation=45)
plt.show()
```

### MULTIVARIATE ANALYSIS

```python
plt.figure(figsize=(10,5))
sns.boxplot(x="Category",y="Purchase Amount (USD)",hue="Gender",data=pro)
plt.title("Category vs Purchase Amount by Gender")
plt.xticks(rotation=45)
plt.show()
```

### Interpretation
#### - Spending is almost the same for both genders in every category
#### - The small differences change by category
#### - All categories have similar spending range

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x="Age",y="Purchase Amount (USD)",hue="Gender",data=pro)
plt.title("Age vs Purchase Amount by Gender")
plt.xlabel("Age")
plt.ylabel("Purchase Amount (USD)")
plt.show()
```

### Interpretation
#### - There is no clear pattern between age and spending
#### - Males and Females are mixed togather everywhere 
#### - Customers of all ages can be big or small spenders

```python
plt.figure(figsize=(8,5))
sns.barplot(x="Subscription Status",y="Purchase Amount (USD)",hue="Discount Applied",data=pro)
plt.title("Subscription Status vs Purchase Amount by Discount Applied")
plt.xticks(rotation=45)
plt.show()
```

### Interpretation
#### - Average spending is almost the same in every group
#### - Subscribed customers only appear wuth a discount
#### - Discounts and subscription do not make customers spend more

```python
plt.figure(figsize=(8,5))
sns.heatmap(pro.pivot_table(values="Purchase Amount (USD)",index="Category",columns="Season",aggfunc="mean"),annot=True,fmt="1f",cmap="Blues")
plt.title("Avearge Purcahse Amount:Category vs Season")
plt.show()
```

### Interpretation
#### - Fall is the best season for sales
#### - Outerwear spends the least in most seasons
#### - The differences are small
